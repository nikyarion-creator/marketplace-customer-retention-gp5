# Групповой проект №5. Deep Learning

## Customer Retention DSS для e-commerce

**Идея проекта:** построить прототип системы поддержки решений, которая выявляет клиентов с высоким риском ухода и помогает сформировать для них персональный оффер.

**Компания / платформа:** e-commerce / маркетплейс на данных Olist.

**Общая бизнес-цель:** снизить отток клиентов и повысить повторные покупки за счет персонализированных предложений.

## 15. Постановка текстовой задачи

Используем `review_comment_message`.

Вариант таргета:

- review_score == 3 → neutral;
- review_score >= 4 → positive.

Бизнес-смысл: текстовая модель помогает понять настроение клиента и выбрать тип оффера.
